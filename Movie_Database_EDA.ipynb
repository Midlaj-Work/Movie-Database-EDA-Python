```python
# =====================================================
# STEP 2 : Load Dataset
# =====================================================


df = pd.read_csv(
    r"C:\Users\MIDLAJ\OneDrive\Documents\DA03.PDF\Python Projects\mymoviedb (1).csv",
    engine="python"
)

# Display the first 5 rows

df.head()
```

```python
# =====================================================
# STEP 4 : Data Cleaning
# =====================================================

# Remove Unnecessary Columns

df.drop(columns=["Overview", "Poster_Url"], inplace=True)

df.head()
```

```python

# Check Missing Values

print(df.isnull().sum())
```

```python

# Remove Missing Values

df.dropna(inplace=True)
print(df.isnull().sum())
```

```python

# Convert Data Types

df["Release_Date"] = pd.to_datetime(df["Release_Date"])
df["Vote_Count"] = df["Vote_Count"].astype(int)
df["Vote_Average"] = df["Vote_Average"].astype(float)
```

```python

# Verify Data Types

df.info()
```

```python

# Check Final Dataset Shape

print(df.shape)
```

```python
# =====================================================
# STEP 5 : Feature Engineering
# =====================================================

# Create Release Year

df["Release_Year"] = df["Release_Date"].dt.year
df.head()
```

```python

#Create Release Month

df["Release_Month"] = df["Release_Date"].dt.month
df.head()
```

```python

# Verify New Columns

print(df[["Release_Date", "Release_Year", "Release_Month"]].head())
```

```python

# Check Updated Dataset Shape

print(df.shape)
```

```python

# =====================================================
# STEP 6 : Exploratory Data Analysis (EDA)
# =====================================================

# Top 10 Movie Genres


genre_count = df["Genre"].value_counts().head(10)
print(genre_count)
```

```python

# Visualize Top 10 Movie Genres

plt.figure(figsize=(10,6))
sns.barplot(
    x=genre_count.values,
    y=genre_count.index
)

plt.title("Top 10 Movie Genres")
plt.xlabel("Number of Movies")
plt.ylabel("Genre")
plt.tight_layout()
plt.show()
```

```python

# Top 10 Original Languages

language_count = df["Original_Language"].value_counts().head(10)
print(language_count)
```

```python

# Visualize Top 10 Original Languages

plt.figure(figsize=(10,6))
sns.barplot(
    x=language_count.values,
    y=language_count.index
)

plt.title("Top 10 Original Languages")
plt.xlabel("Number of Movies")
plt.ylabel("Language")
plt.tight_layout()
plt.show()
```

```python

# Top 10 Most Popular Movies

top_popular = df.sort_values(by="Popularity", ascending=False).head(10)
top_popular[["Title", "Popularity"]]
```

```python

# Visualize Top 10 Most Popular Movies

plt.figure(figsize=(12,6))
sns.barplot(
    x="Popularity",
    y="Title",
    data=top_popular
)

plt.title("Top 10 Most Popular Movies")
plt.xlabel("Popularity Score")
plt.ylabel("Movie Title")
plt.tight_layout()
plt.show()
```

```python

# Top 10 Highest Rated Movies

top_rated = df.sort_values(by="Vote_Average", ascending=False).head(10)
top_rated[["Title", "Vote_Average"]]
```

```python

# Visualize Top 10 Highest Rated Movies

plt.figure(figsize=(12,6))
sns.barplot(
    x="Vote_Average",
    y="Title",
    data=top_rated
)

plt.title("Top 10 Highest Rated Movies")
plt.xlabel("Average Rating")
plt.ylabel("Movie Title")
plt.tight_layout()
plt.show()
```

```python

# Movies Released by Year

movies_per_year = df["Release_Year"].value_counts().sort_index()
print(movies_per_year)
```

```python

# Visualize Movies Released by Year

plt.figure(figsize=(12,6))
sns.lineplot(
    x=movies_per_year.index,
    y=movies_per_year.values,
    marker="o"
)

plt.title("Movies Released by Year")
plt.xlabel("Release Year")
plt.ylabel("Number of Movies")
plt.tight_layout()
plt.show()
```
